```python
# 11. Gere um array de 100 inteiros e encontre o valor que mais se repete
import numpy as np
array = np.random.randint(0, 101, 100)
mais_repetido = None    #aqui é criado variáveis para guardar o valor mais repetido
maior_contagem = 0
for numero in array:
    contagem = np.count_nonzero(array == numero)  #aqui é contado quantas vezes esse número aparece
    if contagem > maior_contagem:       # if essa contagem for maior que a anterior, atualiza
        maior_contagem = contagem
        mais_repetido = numero

print("Array:\n", array)
print("Valor que mais se repete:", mais_repetido)
print("Quantidade de vezes:", maior_contagem)
```

```python
#Exercicio 19 | Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado)
import numpy as np
array_1 = np.random.randint(1, 11, 20) # aqui é geado dois arrays inteiros aleatórios
array_2 = np.random.randint(1, 11, 20)
produto_escalar = 0     # o prduto_escalar multiplica posição a posição e soma
for i in range(20):
    produto_escalar += array_1[i] * array_2[i]
norma_1 = np.sqrt(np.sum(array_1**2))  # aqui é feita a raiz da soma dos quadrados dos dois arrays
norma_2 = np.sqrt(np.sum(array_2**2))
cosseno = produto_escalar / (norma_1 * norma_2)  #calcula o cosseno do ângulo
angulo = np.degrees(np.arccos(cosseno))  # o angulo é convertido para graus
print("Araay 1:", array_1)
print("Array 2:", array_2)
print("Ângulo entre os vetores:", angulo, "graus")
```

# **Exercício de SciPy**

```python
# 1. Calcule a integral da funcão f(x) = x2 sin(x) no intervalo de 0 a π.
import numpy as np
from scipy.integrate import quad
def f(x):
	return x**2 * np.sin(x)     # aqui usamos a funções que é uma forma curta de criar funções anônimas, ela equivale a escrever a equação
resultado, erro = quad(f, 0, np.pi) # aqui é calculado o valor integral de f(x) no intervalo [0, π], onde quad devolve dois valores
print("Resultado da integral:", resultado) # - resultado: o valor da integral - (valores devolvidos)
print("Erro estimado:", erro)               # - erro: uma estimativa do erro numérico do cálculo -(valores devolvidos)
```

```python
# 2. Resolva a equação diferencial dy dx = x−y, com y(0) = 1, no intervalo de 0a 10.
import numpy as np
from scipy.integrate import solve_ivp
def equacao(x, y):  # Define a equação diferencial y' = x - y.
	return x - y
solucao = solve_ivp(equacao, [0, 10], [1], t_eval=np.linspace(0, 10, 100)) # Resolve a equação começando com y(0) = 1.
print("Valores de x:", solucao.t)
print("Valores de y:", solucao.y[0])
```

```python
# 3. Encontre as raízes da função f(x) = x4 − 3x3 + 2 utilizando SciPy.
from scipy.optimize import root
def f(x):
    return x ** 4 - 3 * x ** 3 + 2
raiz_1 = root(f, 1)   # procura uma raiz partindo de 1.
raiz_2 = root(f, 3)    # procura outra raiz partindo de 3.
print("Raiz 1:", raiz_1.x)
print("Raiz 2:", raiz_2.x)
```

```python
# 4. Use a função de otimização de SciPy para maximizar f(x) = −x2+4x+1.
from scipy.optimize import minimize      # minimize trabalha com mínimos, por isso usamos o negativo da função.
def f(x):
	return -(-x[0]**2 + 4*x[0] + 1)
resultado = minimize(f, [0])    # procura o melhor valor começando em x = 0.
print("Valor de x:", resultado.x)
print("Máximo encontrado:", -resultado.fun)
```

```python
# 5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).
import numpy as np
from scipy.fft import fft
t = np.linspace(0, 1, 500, endpoint=False)  # aqui é criado os valores do tempo e o sinal.
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)
transformada = fft(sinal)  # o calculo da transformada de Fourier é feito.
print("Transformada de Fourier:", transformada)
```

```python
# 6. Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.
import numpy as np
from scipy.linalg import inv
matriz = np.random.rand(5, 5) + np.eye(5)   # cria uma matriz 5x5 e soma a identidade para facilitar a inversão.
matriz_inversa = inv(matriz)  # a matriz inversa é calculada.
print("Matriz original:\n", matriz)
print("Matriz inversa:\n", matriz_inversa)
```

```python
# 7. Use a função de interpolação cúbica de SciPy para interpolar os dados(0, 0), (1,2),(2,3),(3,5),(4,4).
import numpy as np
from scipy.interpolate import interp1d
x = np.array([0, 1, 2, 3, 4])       # guarda os pontos conhecidos de x e y.
y = np.array([0, 2, 3, 5, 4])
interpolacao = interp1d(x, y, kind="cubic")  # Cria uma interpolação cúbica.
novos_x = np.array([0.5, 1.5, 2.5, 3.5])  # calcula valores entre os pontos conhecidos.
print("Valores interpolados:", interpolacao(novos_x))
```

```python
# 8. Resolva o sistema de equações 3x + 2y = 5 e x+4y = 6.
import numpy as np
from scipy.linalg import solve
A = np.array([[3, 2], [1, 4]])  # aqui é montado a matriz dos coeficientes e o vetor dos resultados.
B = np.array([5, 6])
solucao = solve(A, B)    #o sistema resolve na forma A vezes x igual a B.
print("Solução [x, y]:", solucao)
```

```python
# 9. Ajuste uma curva quadrática aos pontos (1,2),(2,4),(3,5),(4,7),(5,8).
import numpy as np
from scipy.optimize import curve_fit
x = np.array([1, 2, 3, 4, 5])   # guarda os pontos usados no ajuste.
y = np.array([2, 4, 5, 7, 8])
def parabola(x, a, b, c):   # define a curva quadrática
	return a*x**2 + b*x + c
parametros, _ = curve_fit(parabola, x, y)  # encontra os valores de a, b e c que melhor se ajustam aos pontos.
print("Parâmetros [a, b, c]:", parametros)
```

```python
# 10. Calcule a integral dupla da função f(x,y) = xy + x2 no intervalo [0,2]
# para x e [0,1] para y.
from scipy.integrate import dblquad
def f(y, x): # a função recebe primeiro y e depois x
	return x*y + x**2.
def limite_inferior(x): # aqui é definido os limites de y
	return 0
def limite_superior(x):
	return 1
resultado, erro = dblquad(f, 0, 2, limite_inferior, limite_superior) # integra x de 0 a 2 e y de 0 a 1.
print("Resultado da integral dupla:", resultado)
print("Erro estimado:", erro)
```

```python
# 11. Encontre os autovalores e autovetores de uma matriz 4x4 de números inteiros aleatórios.
import numpy as np
from scipy.linalg import eig  # essa lib já responsavel para calcular os autovalores e autovetores
matriz = np.random.randint(1, 10, (4, 4)) # cria uma matriz 4x4 com números inteiros.
autovalores, autovetores = eig(matriz)  # aqui é calculado os autos
print("Matriz:\n", matriz)
print("Autovalores:\n", autovalores)
print("Autovetores:\n", autovetores)
```

```python
# 12. Resolva a equação ex − x2 = 0 para encontrar a raiz mais próxima de x =1.
import numpy as np
from scipy.optimize import fsolve
def equacao(x): # Define a equação e começa a procura em x = 1
	return np.exp(x) - x**2
raiz = fsolve(equacao, 1)
print("Raiz encontrada:", raiz)
```

```python
# 13. Calcule o determinante de uma matriz 6x6 de números aleatórios entre-10 e 10.
import numpy as np
from scipy.linalg import det
matriz = np.random.randint(-10, 11, (6, 6))  # gera uma matriz 6x6 com valores entre -10 e 10.
determinante = det(matriz)   #variavel do determinante da matriz.
print("Matriz:\n", matriz)
print("Determinante:", determinante)
```

```python
# 14. Utilize a função de otimização multivariada para minimizar f(x,y) = (x− 2)2 +(y −3)2.
from scipy.optimize import minimize
def f(valores):  # define a função com duas variáveis.
	x, y = valores
	return (x - 2)**2 + (y - 3)**2
resultado = minimize(f, [0, 0])  # Dcomeça a busca pelo 0
print("Ponto do mínimo [x, y]:", resultado.x)
print("Valor mínimo:", resultado.fun)
```

```python
# 15. Calcule a transformada de Fourier inversa de um sinal complexo gerado aleatoriamente.
import numpy as np
from scipy.fft import ifft
sinal_complexo = np.random.rand(10) + 1j*np.random.rand(10)  # cria um sinal complexo com parte real e imaginária.
resultado = ifft(sinal_complexo)  # calcula a transformada inversa.
print("Sinal complexo:", sinal_complexo)
print("Transformada inversa:", resultado)
```

```python
# 16. Resolva a equação diferencial de segunda ordem d2y dx2 +y = 0, comcondições iniciais y(0) = 1 e y′(0) = 0.
import numpy as np
from scipy.integrate import solve_ivp
def equacao(x, estado):  # transforma a equação de segunda ordem em duas de primeira ordem.
	y, derivada = estado
	return [derivada, -y] # estado[0] é y e estado[1] é y'.
solucao = solve_ivp(equacao, [0, 2*np.pi], [1, 0], t_eval=np.linspace(0, 2*np.pi, 100))
print("Valores de y:", solucao.y[0])
```

```python
# 17. Encontre a raiz cúbica de 27 utilizando SciPy.
from scipy.optimize import fsolve
def equacao(x): # procura o valor que torna x³ - 27 igual a zero.
	return x**3 - 27
raiz = fsolve(equacao, 2)
print("Raiz cúbica de 27:", raiz)
```

```python
# 18. Interpole os valores de uma função f(x) = x2 nos pontos [1,2,3,4,5]
# utilizando spline cúbica.
import numpy as np
from scipy.interpolate import CubicSpline
x = np.array([1, 2, 3, 4, 5])  # cria os pontos da função f(x) = x².
y = x**2
spline = CubicSpline(x, y)  # cria a spline cúbica.
novos_x = np.array([1.5, 2.5, 3.5, 4.5])
print("Valores interpolados:", spline(novos_x))
```

```python
#19. Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.
import numpy as np
from scipy.fft import fft
amostras = np.arange(32)  # cria um sinal discreto formado por seno e cosseno.
sinal = np.sin(2*np.pi*amostras/8) + np.cos(2*np.pi*amostras/4)
transformada = fft(sinal)  # calcula a transformada de Fourier do sinal.
print("Transformada de Fourier:\n", transformada)
```

```python
# 14. Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.
import pandas as pd
df_vendas = pd.read_excel("/PANDAS_DADOS/vendas.xlsx")
df_lojas = pd.read_excel("/PANDAS_DADOS/lojas.xlsx")
df_mesclado = pd.merge(df_vendas, df_lojas, on="ID", how="inner")  # junta os dois DataFrames pela coluna ID
print(df_mesclado[["ID", "Produto", "Loja", "Vendedor"]])
```

```python
# 15. Crie um DataFrame com dados faltantes e use a interpolação linear para preenchê-los.
import pandas as pd
import numpy as np

df_alimentos = pd.DataFrame({
    "Produto": ["Arroz", "Feijao", "Acucar", "Cafe", "Oleo"],
    "Preco": [22.90, np.nan, 4.20, np.nan, 9.80]  # valores faltantes propositais
})
print("Antes da interpolacao:")
print(df_alimentos)

df_alimentos["Preco"] = df_alimentos["Preco"].interpolate(method="linear")  # interpola os valores faltantes
print("\nDepois da interpolacao:")
print(df_alimentos)
```

```python
# 16. Crie um gráfico de barras a partir de um DataFrame usando a biblioteca Pandas.
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

df_vendas = pd.read_excel("/PANDAS_DADOS/vendas.xlsx")
total_por_categoria = df_vendas.groupby("Categoria")["Quantidade"].sum()  # agrupa para o grafico

grafico = total_por_categoria.plot(kind="bar", figsize=(8, 5)) # define tamanho
grafico.set_title("Quantidade vendida por categoria") # define titulo
grafico.set_xlabel("Categoria")  #eixo x
grafico.set_ylabel("Quantidade")  #eixo y
plt.xticks(rotation=45)
plt.tight_layout()

display(grafico.figure)
```

```python
# 17. Adicione uma coluna ao DataFrame que calcule o imposto (5%) sobre o valor da coluna Total.
import pandas as pd

df_vendas = pd.read_excel("/PANDAS_DADOS/vendas.xlsx")
df_vendas["Total"] = df_vendas["Preco"] * df_vendas["Quantidade"]  # coluna Total do exercicio 7
df_vendas["Imposto"] = df_vendas["Total"] * 0.05  # 5% sobre o Total
print(df_vendas[["Produto", "Total", "Imposto"]])
```

```python
# 18. Exiba apenas as linhas de um DataFrame onde o nome do produto começa com a letra "B".
import pandas as pd

df_vendas = pd.read_excel("/PANDAS_DADOS/vendas.xlsx")
df_letra_b = df_vendas[df_vendas["Produto"].str.startswith("B")]  # filtra produtos que comecam com "B"
print(df_letra_b[["Produto", "Categoria"]])
```

```python
# 19. Crie um DataFrame a partir de um dicionário de listas e calcule a média de cada coluna numérica.
import pandas as pd
dados = {
    "Produto": ["Arroz", "Feijao", "Acucar", "Cafe", "Oleo de Soja"],
    "Preco": [22.90, 8.50, 4.20, 18.90, 9.80],
    "Quantidade": [10, 15, 20, 8, 12]
}
df_alimentos = pd.DataFrame(dados)  # cria o DataFrame a partir do dicionario de listas
media_colunas = df_alimentos.select_dtypes("number").mean().reset_index() # calcula a media das colunas numericas
media_colunas.columns = ["Coluna", "Media"]
print(media_colunas)
```

```python
# 20. Agrupe os dados por uma coluna categórica e calcule a mediana de uma coluna numérica.
import pandas as pd

df_vendas = pd.read_excel("/PANDAS_DADOS/vendas.xlsx")
mediana_por_categoria = df_vendas.groupby("Categoria")["Preco"].median()  # mediana do Preco por Categoria
print(mediana_por_categoria)
```

# **Exercícios Matplotlib**

```python
# 1. Crie um gráfico de linha para a função y = x3 no intervalo de x = −10 até x = 10.
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-10, 10, 200)  # cria os valores de x e calcula y = x³.
y = x**3
plt.plot(x, y, color="#2A6F97")  # cria o gráfico de linha.
plt.title("Função y = x³")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 2. Plote um gráfico de dispersão para 100 pontos aleatórios com uma paleta de cores baseada nos valores de y.
import matplotlib.pyplot as plt
import numpy as np
x = np.random.rand(100)  # gera 100 pontos aleatórios.
y = np.random.rand(100)
plt.scatter(x, y, c=y, cmap="viridis", alpha=0.8)  # usa os valores de y para escolher as cores dos pontos.
plt.colorbar(label="Valor de y")
plt.title("Dispersão com cores baseadas em y")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# 3. Gere um histograma para 500 números inteiros gerados aleatoriamente entre 0 e 100.
import matplotlib.pyplot as plt
import numpy as np
dados = np.random.randint(0, 101, 500)  # gera 500 números inteiros entre 0 e 100.
plt.hist(dados, bins=10, color="#5B8E7D", edgecolor="white")  # cria o histograma.
plt.title("Histograma de números inteiros")
plt.xlabel("Número")
plt.ylabel("Frequência")
plt.show()
```

```python
# 4. Crie um gráfico de barras para 5 categorias com valores diferentes, colorindo cada barra de uma cor diferente.
import matplotlib.pyplot as plt
categorias = ["A", "B", "C", "D", "E"]  # define as categorias e seus valores.
valores = [4, 7, 2, 9, 5]
cores = ["#4C78A8", "#F58518", "#54A24B", "#E45756", "#72B7B2"]
plt.bar(categorias, valores, color=cores) # cria uma barra para cada categoria.
plt.title("Valores por categoria")
plt.xlabel("Categoria")
plt.ylabel("Valor")
plt.show()
```

```python
# 5. Faça um gráfico de pizza para representar as porcentagens de quatro setores de mercado com rótulos personalizados.
import matplotlib.pyplot as plt
rotulos = ["Alimentação", "Tecnologia", "Saúde", "Transporte"]  # define setores de mercado e suas participacoes.
valores = [35, 30, 20, 15]
cores = ["#4C78A8", "#F2CF5B", "#59A14F", "#E15759"]
plt.pie(valores, labels=rotulos, colors=cores, autopct="%1.1f%%", startangle=90)  # cria o gráfico de pizza mostrando as porcentagens.
plt.title("Participação no mercado")
plt.show()
```

```python
# 6. Plote dois gráficos de linha em subplots diferentes: y1 = ex e y2 = log(x).
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0.1, 3, 100) # usa x positivo porque log(x) não existe para x igual a zero.
y1 = np.exp(x)
y2 = np.log(x)
plt.subplot(2, 1, 1)  # cria dois gráficos na mesma figura.
plt.plot(x, y1, color="#2A6F97")
plt.title("y1 = eˣ")
plt.subplot(2, 1, 2)
plt.plot(x, y2, color="#E07A5F")
plt.title("y2 = log(x)")
plt.tight_layout()
plt.show()
```

```python
# 7. Crie um gráfico de dispersão com tamanho dos pontos variando de acordo com os valores de y.
import matplotlib.pyplot as plt
import numpy as np
x = np.random.rand(100) # gera os pontos.
y = np.random.rand(100)
plt.scatter(x, y, s=y * 300 + 10, color="#7A5195", alpha=0.7) # usa y para controlar o tamanho dos pontos.
plt.title("Tamanho dos pontos baseado em y")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# 8. Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2 * np.pi, 100) # cria a função seno e uma margem para o intervalo.
y = np.sin(x)
margem = 0.2
plt.plot(x, y, color="#2A6F97", label="sen(x)") # desenha a linha e o sombreado do intervalo.
plt.fill_between(x, y - margem, y + margem, color="#A8DADC", alpha=0.6, label="Intervalo")
plt.title("Seno com intervalo de confiança")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# 9. Adicione título, rótulos e uma grade personalizada a um gráfico de linha.
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100) # cria os valores do gráfico
y = np.sin(x)
plt.plot(x, y, color="#E07A5F")
plt.title("Gráfico de uma função seno")
plt.xlabel("Eixo x")
plt.ylabel("Eixo y")
plt.grid(color="#9CA3AF", linestyle="--", linewidth=0.7, alpha=0.6)
plt.show()
```

```python
# 10. Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100) # cria três funções para comparar.
plt.plot(x, np.sin(x), color="#2A6F97", linestyle="-", label="sen(x)")
plt.plot(x, np.cos(x), color="#E07A5F", linestyle="--", label="cos(x)")
plt.plot(x, np.sin(x) + np.cos(x), color="#5B8E7D", linestyle=":", label="sen(x) + cos(x)")
plt.title("Três gráficos de linha")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# 11. Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.
import matplotlib.pyplot as plt
import numpy as np
categorias = ["A", "B", "C"] #define os valores dos três grupos.
grupo_1 = np.array([3, 5, 4])
grupo_2 = np.array([2, 3, 1])
grupo_3 = np.array([1, 2, 3])
plt.bar(categorias, grupo_1, color="#4C78A8", label="Grupo 1")  # empilha um grupo sobre o outro usando bottom.
plt.bar(categorias, grupo_2, bottom=grupo_1, color="#F2CF5B", label="Grupo 2")
plt.bar(categorias, grupo_3, bottom=grupo_1 + grupo_2, color="#59A14F", label="Grupo 3")
plt.title("Barras empilhadas")
plt.xlabel("Categoria")
plt.ylabel("Valor")
plt.legend()
plt.show()
```

```python
# 12. Plote um gráfico de linha com uma função quadrática e ajuste os limites do eixo para focar nos valores x = −5 a x = 5.
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-10, 10, 200) # cia a função quadrática y = x².
y = x**2
plt.plot(x, y, color="#4C78A8")
plt.xlim(-5, 5) # limita a visualização do eixo x entre -5 e 5.
plt.title("Função quadrática")
plt.xlabel("x")
plt.ylabel("y = x²")
plt.show()
```

```python
# 13. Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando Matplotlib.
import matplotlib.pyplot as plt
import numpy as np
x = np.random.rand(100) # gera os valores dos três eixos.
y = np.random.rand(100)
z = np.random.rand(100)
figura = plt.figure()
eixo = figura.add_subplot(111, projection="3d") #grafico 3d é criado
eixo.scatter(x, y, z, color="#2A9D8F", alpha=0.8)
eixo.set_title("Dispersão 3D")
eixo.set_xlabel("x")
eixo.set_ylabel("y")
eixo.set_zlabel("z")
plt.show()
```

```python
# 18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.
import matplotlib.pyplot as plt
categorias = ["A", "B", "C", "D", "E"] # define valores positivos e negativos.
valores = [-4, 3, -2, 6, -1]
cores = ["#E15759" if valor < 0 else "#59A14F" for valor in valores]
plt.bar(categorias, valores, color=cores) # cria as barras e uma linha no zero.
plt.axhline(0, color="#444444", linewidth=0.8)
plt.title("Valores positivos e negativos")
plt.xlabel("Categoria")
plt.ylabel("Valor")
plt.show()
```

```python
# 19. Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)  # cria duas séries de dados.
serie_1 = np.sin(x)
serie_2 = np.cos(x)
plt.plot(x, serie_1, color="#2A6F97", label="Seno") #plota as series
plt.plot(x, serie_2, color="#E07A5F", label="Cosseno")
plt.title("Duas séries de dados")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# 20. Adicione uma grade com espaçamento personalizado a um gráfico de linha.
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2 * np.pi, 100) # cria os valores do gráfico.
y = np.sin(x)
plt.plot(x, y, color="#7A5195") # define a distância dos pontos dos eixos e mostra a grade.
plt.xticks(np.arange(0, 7, 1))
plt.yticks(np.arange(-1, 1.1, 0.5))
plt.grid(True, linestyle=":", color="#9CA3AF", alpha=0.7)
plt.title("Gráfico com grade personalizada")
plt.xlabel("x")
plt.ylabel("sen(x)")
plt.show()
```
